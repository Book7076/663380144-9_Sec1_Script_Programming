```python
#  Block A: Evaluation Check
temp = 25
is_raining = True

if temp > 20 and not is_raining:
    print("Perfect day for a picnic!")
elif temp <= 20 or is_raining:
    print("Stay indoors or bring an umbrella.") # Expected output matches here
else:
    print("Unsure about the weather.")
```

ตอบ
ลำดับการทำงาน:
 1. ตรวจสอบ if: temp > 20 (True) and not is_raining (False) -> ได้ False (ข้าม)
 2. ตรวจสอบ elif: temp <= 20 (False) or is_raining (True) -> ได้ True (ทำงานในบล็อกนี้)
 ผลลัพธ์ (Output): Stay indoors or bring an umbrella.

```python
#  Block B: Vectorized Filter
phrase = "Python 2026 Engine"
consonants = {c.lower() for c in phrase if c.isalpha() and c.lower() not in 'aeiou'}
print(f"Unique consonants: {consonants}")
```

 ตอบ
 ลำดับการทำงาน:
1.เช็กว่าเป็นตัวอักษรหรือไม่ (c.isalpha()) ถ้าเจอช่องว่างหรือตัวเลข จะได้ False และข้ามตัวนั้นทันที

2.เช็กว่าไม่ใช่สระใช่หรือไม่ (c.lower() not in 'aeiou') ถ้าตัวอักษรนั้นเป็น สระ (a, e, i, o, u) จะได้ False

3.จัดเก็บผลลัพธ์ ตัวอักษรที่ผ่านทั้ง 2 เงื่อนไข (เป็นพยัญชนะ) จะถูกแปลงเป็นตัวพิมพ์เล็กแล้วเก็บเข้ากลุ่ม (Set) ซึ่งมีคุณสมบัติคัดตัวที่ซ้ำออกโดยอัตโนมัติ

ผลลัพธ์ (Output): ตัวอักษรที่ผ่านเงื่อนไขคือพยัญชนะภาษาอังกฤษ ได้แก่ p, y, t, h, n, g

```python
# ฉากเริ่มต้น
print("คุณเจอประตูลึกลับ และมีแมวนั่งอยู่")
print("1 = เข้าประตู")
print("2 = กลับบ้าน")

# รับค่าจากผู้เล่น
choice1 = input("เลือกข้อไหนดี (1 หรือ 2): ")

# เงื่อนไขหลัก
if choice1 == "1":
    print("\nคุณเข้ามาในป่าเวทมนตร์แล้วเจอทางแยก")
    print("A = ไปที่ถ้ำ")
    print("B = ไปที่สะพาน")

    choice2 = input("เลือกทางไหนดี (A หรือ B): ")

    # เงื่อนไขย่อยข้างใน
    if choice2 == "A":
        print("ยินดีด้วย! คุณเจอขุมทรัพย์ทองคำ")
    elif choice2 == "B":
        print("ตกสะพาน! แต่นกฮูกบินมาช่วยไว้ทัน")
    else:
        print("พิมพ์ผิด แมวเดินหนีเลย!")

elif choice1 == "2":
    print("\nคุณเดินกลับบ้านไปนอนกินหมูกระทะสบายใจ")

else:
    print("\nพิมพ์ไม่ถูกต้อง เกมจบลงทันที")
```

START

1. แสดงข้อความฉากเริ่มต้น: "คุณเจอประตูลึกลับ และมีแมวนั่งอยู่"
2. แสดงทางเลือก: "1 = เข้าประตู", "2 = กลับบ้าน"
3. INPUT() -> รับค่าจากผู้เล่นมาเก็บไว้ในตัวแปร choice1

4. ตรวจสอบเงื่อนไขหลัก (IF-ELIF-ELSE ของ choice1):

    • IF choice1 เท่ากับ "1":
        - แสดงข้อความ: "คุณเข้ามาในป่าเวทมนตร์แล้วเจอทางแยก"
        - แสดงทางเลือกย่อย: "A = ไปที่ถ้ำ", "B = ไปที่สะพาน"
        - INPUT() -> รับค่าจากผู้เล่นมาเก็บไว้ในตัวแปร choice2
        
        * ตรวจสอบเงื่อนไขย่อย (IF-ELIF-ELSE ของ choice2):
            - IF choice2 เท่ากับ "A":
                แสดงข้อความ: "ยินดีด้วย! คุณเจอขุมทรัพย์ทองคำ"
            - ELIF choice2 เท่ากับ "B":
                แสดงข้อความ: "ตกสะพาน! แต่นกฮูกบินมาช่วยไว้ทัน"
            - ELSE (พิมพ์คำอื่น):
                แสดงข้อความ: "พิมพ์ผิด แมวเดินหนีเลย!"

    • ELIF choice1 เท่ากับ "2":
        - แสดงข้อความ: "คุณเดินกลับบ้านไปนอนกินหมูกระทะสบายใจ"

    • ELSE (พิมพ์คำอื่น):
        - แสดงข้อความ: "พิมพ์ไม่ถูกต้อง เกมจบลงทันที"
