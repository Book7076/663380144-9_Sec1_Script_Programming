```python
sentence = input("กรุณาใส่ประโยค: ")

sentence = sentence.lower()

words = sentence.split()

word_counts = {}

for word in words:
    word = word.strip('.,!?"')

    if word in word_counts:
        word_counts[word] += 1
    else:
        word_counts[word] = 1

print("จำนวนความถี่ของแต่ละคำ:")
print(word_counts)
```
