3.ประกาศตัวแปรและคำนวณ

```python
num1 = 6
num2 = 7

Plus = num1+num2
Minus = num1-num2
Divide = num1/num2
Multiply = num1*num2

print("Plus =",Plus)
print("Minus =",Minus)
print("Divide =",Divide)
print("Multiply =",Multiply)
```
