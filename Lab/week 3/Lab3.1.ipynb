Part 1: ตารางสูตรคูณแม่เดียว (Single Number Table):

```python
n = int(input("Enter a number for its multiplication table (e.g., 7): "))

for i in range(1, 13):
    print(f"{n} * {i} = {n * i}")

print()
```

Part 2: ตารางสูตรคูณแบบเต็ม (Full Multiplication Grid) - โจทย์ท้าทาย/ส่วนขยาย:

```python
print("12x12 Multiplication Table:")

for i in range(1, 13):
    for j in range(1, 13):
        print(f"{i*j:4}", end="")
    print()

print()
```
