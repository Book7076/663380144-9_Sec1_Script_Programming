Activity 4.1: List vs. Tuple: Scenario Matchin

1. ข้อมูลอุณหภูมิประจำวันที่มีการบันทึกไว้ตลอดหนึ่งสัปดาห์

ตอบ
ใช้ list
เหตุผล: ข้อมูลอาจมีการแก้ไขหรืออัปเดตได้

2. หากบันทึกอุณหภูมิผิดหรือมีการเพิ่มข้อมูลค่าสี RGB ของเฉดสีเฉพาะเฉดหนึ่ง (เช่น สีแดงคือ (255, 0, 0))

ตอบ ใช้ tuple
เหตุผล: ค่าสี RGB เป็นค่าคงที่ ไม่ควรเปลี่ยนแปลงหลังจากกำหนด

3. รายการสินค้าในรถเข็นชอปปิงที่สามารถกดเพิ่มหรือลบออกได้

ตอบ ใช้ list
เหตุผล: สามารถเพิ่ม ลบ หรือแก้ไขรายการสินค้าได้ตลอดเวลา

4. รายชื่อวันในหนึ่งสัปดาห์เรียงตามลำดับ (จันทร์ - อาทิตย์)

ตอบ ใช้ tuple
เหตุผล: ชื่อวันและลำดับของวันเป็นข้อมูลคงที่ ไม่เปลี่ยนแปลง

5. คะแนนเก็บของนักเรียนจากการส่งงานชุดต่าง ๆ

ตอบ ซึ่งอาจมีการอัปเดตหรือแก้ไขคะแนนในภายหลังได้
ใช้ list
เหตุผล: คะแนนสามารถแก้ไขหรืออัปเดตได้เมื่อมีการตรวจใหม่หรือเพิ่มคะแนน

6. พิกัด GPS ของสถานที่สำคัญหรือจุดแลนด์มาร์กที่ตั้งอยู่กับที่

ตอบ ใช้ tuple
เหตุผล: พิกัดของสถานที่เป็นข้อมูลคงที่ ไม่ควรเปลี่ยนแปลง จึงเหมาะกับ tuple เช่น (16.4322, 102.8236)

Activity 4.2: Interactive To-Do List (using List Operations)

```python
print("=== To-Do List Program ===")

todo_list = []

while True:
    print("\n===== MENU =====")
    print("1. Add a task")
    print("2. View tasks")
    print("3. Complete a task")
    print("4. Exit")

    choice = input("Choose an option (1-4): ")

    if choice == "1":
        task = input("Enter a task: ")
        todo_list.append(task)
        print(f"'{task}' has been added.")

    elif choice == "2":
        if len(todo_list) == 0:
            print("No tasks in the to-do list.")
        else:
            print("\nTo-Do List:")
            for i, task in enumerate(todo_list):
                print(f"{i}. {task}")

    elif choice == "3":
        if len(todo_list) == 0:
            print("No tasks to complete.")
        else:
            print("\nCurrent Tasks:")
            for i, task in enumerate(todo_list):
                print(f"{i}. {task}")

            index = int(input("Enter the index of the completed task: "))

            if 0 <= index < len(todo_list):
                completed = todo_list.pop(index)
                print(f"'{completed}' has been completed and removed.")
            else:
                print("Invalid index.")

    elif choice == "4":
        print("Goodbye!")
        break

    else:
        print("Invalid choice. Please try again.")
```
